## Phase 2 — Prepare separate cleaned datasets

```python

# Preserve the original DataFrames
# Create separate working copies for cleaning

transactions_clean = transactions.copy(deep=True)
customers_clean = customers.copy(deep=True)
products_clean = products.copy(deep=True)

print("Working copies created successfully.")

print("Raw transaction rows:", len(transactions))
print("Working transaction rows:", len(transactions_clean))

print("Raw customer rows:", len(customers))
print("Working customer rows:", len(customers_clean))

print("Raw product rows:", len(products))
print("Working product rows:", len(products_clean))
```

```python

# Standardize whitespace and capitalization in transaction text fields
transaction_text_columns = [
    "Sales_Channel",
    "Category",
    "Payment_Method",
    "Transaction_Status"
]

for col in transaction_text_columns:
    transactions_clean[col] = (
        transactions_clean[col]
        .str.strip()
        .str.casefold()
    )

# Standardize customer text fields
customer_text_columns = [
    "Customer_Name",
    "Gender",
    "City",
    "State",
    "Customer_Segment",
    "Loyalty_Tier",
    "Consent_for_Offers"
]

for col in customer_text_columns:
    customers_clean[col] = (
        customers_clean[col]
        .str.strip()
        .str.casefold()
    )

# Standardize product text fields
product_text_columns = ["Product_Name", "Category"]

for col in product_text_columns:
    products_clean[col] = (
        products_clean[col]
        .str.strip()
        .str.casefold()
    )

print("Text formatting standardized.")

print("\nTransaction statuses:")
print(transactions_clean["Transaction_Status"].value_counts(dropna=False))

print("\nSales channels:")
print(transactions_clean["Sales_Channel"].value_counts(dropna=False))

print("\nPayment methods:")
print(transactions_clean["Payment_Method"].value_counts(dropna=False))
```

```python

# Map inconsistent status labels to canonical values
status_mapping = {
    "completed": "completed",
    "complete": "completed",
    "cancelled": "cancelled",
    "returned": "returned",
    "return": "returned",
    "pending": "pending"
}

transactions_clean["Transaction_Status"] = (
    transactions_clean["Transaction_Status"]
    .map(status_mapping)
)

print("Standardized transaction statuses:")
print(
    transactions_clean["Transaction_Status"]
    .value_counts(dropna=False)
)

print(
    "\nTotal transaction rows:",
    len(transactions_clean)
)
```

```python

# Record row count before removing duplicates
rows_before = len(transactions_clean)

# Remove rows that are identical across all columns
transactions_clean = transactions_clean.drop_duplicates()

# Record row count after removing duplicates
rows_after = len(transactions_clean)

print("Rows before deduplication:", rows_before)
print("Rows after deduplication:", rows_after)
print("Duplicate rows removed:", rows_before - rows_after)

# Confirm the original raw data remains unchanged
print("\nOriginal raw transaction rows:", len(transactions))
```

```python

print("Sales channels:")
print(transactions_clean["Sales_Channel"].value_counts(dropna=False))

print("\nPayment methods:")
print(transactions_clean["Payment_Method"].value_counts(dropna=False))

print("\nTransaction categories:")
print(transactions_clean["Category"].value_counts(dropna=False))

print("\nProduct lookup categories:")
print(products_clean["Category"].value_counts(dropna=False))
```

```python

# Join transactions to the product lookup for validation
category_check = transactions_clean.merge(
    products_clean[["Product_ID", "Category"]],
    on="Product_ID",
    how="left",
    suffixes=("_transaction", "_lookup"),
    indicator=True
)

# Count unmatched product IDs
unmatched_products = (category_check["_merge"] != "both").sum()

# Compare transaction categories with lookup categories
category_mismatches = (
    category_check["Category_transaction"]
    != category_check["Category_lookup"]
).sum()

print("Transaction rows checked:", len(category_check))
print("Unmatched product records:", unmatched_products)
print("Category mismatches:", category_mismatches)
```

```python

# Convert the original date strings into proper datetime values
transactions_clean["Transaction_Date_Parsed"] = pd.to_datetime(
    transactions_clean["Transaction_Date"],
    format="%d-%m-%Y",
    errors="coerce"
)

# Count missing dates in the original data
missing_dates = transactions_clean["Transaction_Date"].isna().sum()

# Count invalid, non-missing date strings
invalid_dates = (
    transactions_clean["Transaction_Date"].notna()
    & transactions_clean["Transaction_Date_Parsed"].isna()
).sum()

print("Rows in cleaned transactions:", len(transactions_clean))
print("Missing original dates:", missing_dates)
print("Invalid date strings:", invalid_dates)
print(
    "Successfully parsed dates:",
    transactions_clean["Transaction_Date_Parsed"].notna().sum()
)

print("\nExamples of invalid dates:")
display(
    transactions_clean.loc[
        transactions_clean["Transaction_Date_Parsed"].isna(),
        ["Transaction_ID", "Transaction_Date", "Transaction_Date_Parsed"]
    ].head(10)
)
```

```python
# Find the earliest and latest valid transaction dates
earliest_date = transactions_clean["Transaction_Date_Parsed"].min()
latest_date = transactions_clean["Transaction_Date_Parsed"].max()

print("Earliest valid transaction date:", earliest_date)
print("Latest valid transaction date:", latest_date)

# Count transactions by year and month
monthly_date_counts = (
    transactions_clean
    .dropna(subset=["Transaction_Date_Parsed"])
    .assign(
        Year_Month=lambda df:
            df["Transaction_Date_Parsed"].dt.to_period("M").astype(str)
    )
    .groupby("Year_Month")
    .size()
    .reset_index(name="Transaction_Rows")
)

display(monthly_date_counts)
```

```python

# Count invalid or suspicious quantities
negative_quantity_count = (
    transactions_clean["Quantity"] < 0
).sum()

zero_quantity_count = (
    transactions_clean["Quantity"] == 0
).sum()

# Count invalid or suspicious transaction prices
negative_price_count = (
    transactions_clean["Unit_Price_INR"] < 0
).sum()

zero_price_count = (
    transactions_clean["Unit_Price_INR"] == 0
).sum()

print("Negative quantity rows:", negative_quantity_count)
print("Zero quantity rows:", zero_quantity_count)
print("Negative unit price rows:", negative_price_count)
print("Zero unit price rows:", zero_price_count)

print("\nRows with negative quantities:")
display(
    transactions_clean.loc[
        transactions_clean["Quantity"] < 0,
        [
            "Transaction_ID",
            "Transaction_Status",
            "Quantity",
            "Unit_Price_INR",
            "Discount_Rate"
        ]
    ].head(10)
)

print("\nRows with non-positive unit prices:")
display(
    transactions_clean.loc[
        transactions_clean["Unit_Price_INR"] <= 0,
        [
            "Transaction_ID",
            "Transaction_Status",
            "Quantity",
            "Unit_Price_INR",
            "Discount_Rate"
        ]
    ].head(10)
)
```

```python

# Count discounts outside the expected 0-to-1 range
negative_discount_count = (
    transactions_clean["Discount_Rate"] < 0
).sum()

over_100_discount_count = (
    transactions_clean["Discount_Rate"] > 1
).sum()

exact_100_discount_count = (
    transactions_clean["Discount_Rate"] == 1
).sum()

print("Negative discount rates:", negative_discount_count)
print("Discount rates above 100%:", over_100_discount_count)
print("Discount rates exactly 100%:", exact_100_discount_count)

print("\nRows with invalid discount rates:")
display(
    transactions_clean.loc[
        (transactions_clean["Discount_Rate"] < 0)
        | (transactions_clean["Discount_Rate"] > 1),
        [
            "Transaction_ID",
            "Transaction_Status",
            "Quantity",
            "Unit_Price_INR",
            "Discount_Rate"
        ]
    ]
)
```

```python

# Flag discount values outside the valid range
transactions_clean["Discount_Quality_Flag"] = np.where(
    (transactions_clean["Discount_Rate"] < 0)
    | (transactions_clean["Discount_Rate"] > 1),
    "Invalid",
    "Valid"
)

# Verify the flag
print(
    transactions_clean["Discount_Quality_Flag"]
    .value_counts()
)

print("\nFlagged transactions:")
display(
    transactions_clean.loc[
        transactions_clean["Discount_Quality_Flag"] == "Invalid",
        [
            "Transaction_ID",
            "Discount_Rate",
            "Discount_Quality_Flag"
        ]
    ]
)
```

```python

# Flag invalid quantities
transactions_clean["Quantity_Quality_Flag"] = np.where(
    transactions_clean["Quantity"] < 0,
    "Invalid",
    "Valid"
)

# Flag invalid unit prices
transactions_clean["Price_Quality_Flag"] = np.where(
    transactions_clean["Unit_Price_INR"] <= 0,
    "Invalid",
    "Valid"
)

# Display quality-flag counts
print("Quantity quality:")
print(transactions_clean["Quantity_Quality_Flag"].value_counts())

print("\nPrice quality:")
print(transactions_clean["Price_Quality_Flag"].value_counts())

# Show a combined summary of the three quality checks
print("\nQuality flags summary:")
display(
    transactions_clean[
        [
            "Discount_Quality_Flag",
            "Quantity_Quality_Flag",
            "Price_Quality_Flag"
        ]
    ].apply(lambda column: column.value_counts())
)
```

```python

# Flag rows with at least one invalid quantity, price, or discount
transactions_clean["Has_Critical_Quality_Issue"] = np.where(
    (transactions_clean["Discount_Quality_Flag"] == "Invalid")
    | (transactions_clean["Quantity_Quality_Flag"] == "Invalid")
    | (transactions_clean["Price_Quality_Flag"] == "Invalid"),
    "Yes",
    "No"
)

# Count unique rows affected by at least one issue
print(
    transactions_clean["Has_Critical_Quality_Issue"]
    .value_counts()
)

# Show the affected transaction records
display(
    transactions_clean.loc[
        transactions_clean["Has_Critical_Quality_Issue"] == "Yes",
        [
            "Transaction_ID",
            "Quantity",
            "Unit_Price_INR",
            "Discount_Rate",
            "Discount_Quality_Flag",
            "Quantity_Quality_Flag",
            "Price_Quality_Flag"
        ]
    ].sort_values("Transaction_ID")
)
```

```python

# Review transaction counts and quantities by status
status_summary = (
    transactions_clean
    .groupby("Transaction_Status", dropna=False)
    .agg(
        Transaction_Rows=("Transaction_ID", "size"),
        Unique_Transaction_IDs=("Transaction_ID", "nunique"),
        Total_Quantity=("Quantity", "sum")
    )
    .reset_index()
    .sort_values("Transaction_Rows", ascending=False)
)

display(status_summary)

print(
    "\nTotal rows across statuses:",
    status_summary["Transaction_Rows"].sum()
)
```

```python

print("Status counts in transactions_clean:")
print(
    transactions_clean["Transaction_Status"]
    .value_counts(dropna=False)
)

print("\nStatus counts in the original transactions DataFrame:")
print(
    transactions["Transaction_Status"]
    .value_counts(dropna=False)
)
```

```python

# Compare original and cleaned statuses by transaction ID
status_comparison = transactions[
    ["Transaction_ID", "Transaction_Status"]
].merge(
    transactions_clean[
        ["Transaction_ID", "Transaction_Status"]
    ],
    on="Transaction_ID",
    how="inner",
    suffixes=("_original", "_clean")
)

# Show rows where the original status should map to a different clean status
display(
    status_comparison.loc[
        status_comparison["Transaction_Status_original"].str.strip().str.casefold()
        .isin(["complete", "completed", "cancelled", "return", "returned", "pending"]),
        ["Transaction_ID", "Transaction_Status_original", "Transaction_Status_clean"]
    ].query("Transaction_Status_original != Transaction_Status_clean")
    .head(20)
)

print("Rows in status comparison:", len(status_comparison))
print(
    "Cleaned rows with missing status:",
    transactions_clean["Transaction_Status"].isna().sum()
)
```

```python

# Group transaction statuses for business reporting
status_group_mapping = {
    "completed": "Completed Sale",
    "returned": "Return",
    "cancelled": "Cancelled",
    "pending": "Pending"
}

transactions_clean["Status_KPI_Group"] = (
    transactions_clean["Transaction_Status"]
    .map(status_group_mapping)
)

print("KPI status group counts:")
print(
    transactions_clean["Status_KPI_Group"]
    .value_counts(dropna=False)
)

print(
    "\nRows without a KPI status group:",
    transactions_clean["Status_KPI_Group"].isna().sum()
)
```

```python

# Calculate provisional line value before discount
transactions_clean["Gross_Line_Value_INR"] = (
    transactions_clean["Quantity"]
    * transactions_clean["Unit_Price_INR"]
)

# Summarize completed-sale line values
completed_sales = transactions_clean[
    transactions_clean["Status_KPI_Group"] == "Completed Sale"
]

print("Completed-sale rows:", len(completed_sales))
print(
    "Gross line value before discount (INR):",
    completed_sales["Gross_Line_Value_INR"].sum()
)
print(
    "Rows with negative gross line value:",
    (completed_sales["Gross_Line_Value_INR"] < 0).sum()
)

print("\nSample completed-sale rows:")
display(
    completed_sales[
        [
            "Transaction_ID",
            "Quantity",
            "Unit_Price_INR",
            "Discount_Rate",
            "Gross_Line_Value_INR",
            "Has_Critical_Quality_Issue"
        ]
    ].head(10)
)
```

```python

# Calculate provisional net line value after discount
transactions_clean["Net_Line_Value_INR"] = (
    transactions_clean["Quantity"]
    * transactions_clean["Unit_Price_INR"]
    * (1 - transactions_clean["Discount_Rate"])
)

completed_sales = transactions_clean[
    transactions_clean["Status_KPI_Group"] == "Completed Sale"
]

print(
    "Provisional net line value (INR):",
    round(completed_sales["Net_Line_Value_INR"].sum(), 2)
)

print(
    "Completed-sale rows with negative net value:",
    (completed_sales["Net_Line_Value_INR"] < 0).sum()
)

print(
    "Completed-sale rows with invalid discounts:",
    (completed_sales["Discount_Quality_Flag"] == "Invalid").sum()
)

print("\nRows with negative net line value:")
display(
    completed_sales.loc[
        completed_sales["Net_Line_Value_INR"] < 0,
        [
            "Transaction_ID",
            "Quantity",
            "Unit_Price_INR",
            "Discount_Rate",
            "Net_Line_Value_INR"
        ]
    ]
)
```

```python

# Identify rows eligible for provisional revenue reporting
transactions_clean["Revenue_Eligibility_Flag"] = np.where(
    (transactions_clean["Status_KPI_Group"] == "Completed Sale")
    & (transactions_clean["Quantity"] > 0)
    & (transactions_clean["Unit_Price_INR"] > 0)
    & (transactions_clean["Discount_Rate"].between(0, 1))
    & (transactions_clean["Transaction_Date_Parsed"].notna()),
    "Eligible",
    "Not Eligible"
)

# Summarize eligibility
print(
    transactions_clean["Revenue_Eligibility_Flag"]
    .value_counts(dropna=False)
)

print("\nEligibility by status:")
display(
    transactions_clean.groupby(
        ["Status_KPI_Group", "Revenue_Eligibility_Flag"],
        dropna=False
    ).size().reset_index(name="Rows")
)
```

```python

# Calculate final net line value for eligible revenue rows only
eligible_sales = transactions_clean[
    transactions_clean["Revenue_Eligibility_Flag"] == "Eligible"
]

quality_screened_revenue = eligible_sales["Net_Line_Value_INR"].sum()

print("Eligible sales rows:", len(eligible_sales))
print(
    "Quality-screened net revenue (INR):",
    round(quality_screened_revenue, 2)
)
print(
    "Eligible rows with negative net line value:",
    (eligible_sales["Net_Line_Value_INR"] < 0).sum()
)
print(
    "Eligible rows with missing parsed dates:",
    eligible_sales["Transaction_Date_Parsed"].isna().sum()
)
```

```python
# Step 20: Validate eligible revenue calculation

check_revenue = (
    eligible_sales["Quantity"]
    * eligible_sales["Unit_Price_INR"]
    * (1 - eligible_sales["Discount_Rate"])
).sum()

print("Previously calculated revenue:", round(quality_screened_revenue, 2))
print("Independently recalculated revenue:", round(check_revenue, 2))
print("Revenue calculation matches:", np.isclose(
    quality_screened_revenue,
    check_revenue
))
```

```python
# Step 21: Summarize rows and net value by status

status_summary = (
    transactions_clean
    .groupby("Status_KPI_Group")
    .agg(
        Total_Rows=("Transaction_ID", "size"),
        Eligible_Rows=(
            "Revenue_Eligibility_Flag",
            lambda x: (x == "Eligible").sum()
        ),
        Total_Net_Line_Value_INR=("Net_Line_Value_INR", "sum")
    )
    .reset_index()
)

print(status_summary.to_string(index=False))
```

```python
# Step 22: Monthly quality-screened net revenue

monthly_revenue = (
    eligible_sales
    .assign(
        Month=eligible_sales["Transaction_Date_Parsed"]
        .dt.to_period("M")
        .astype(str)
    )
    .groupby("Month")
    .agg(
        Eligible_Sales_Rows=("Transaction_ID", "size"),
        Net_Revenue_INR=("Net_Line_Value_INR", "sum")
    )
    .reset_index()
    .sort_values("Month")
)

print(monthly_revenue.to_string(index=False))
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick

# Step 23: Plot monthly quality-screened revenue
fig, ax = plt.subplots(figsize=(13, 5))

ax.plot(
    monthly_revenue["Month"],
    monthly_revenue["Net_Revenue_INR"],
    marker="o",
    linewidth=2,
    markersize=5,
    label="Net Revenue"
)

ax.set_title(
    "Monthly Quality-Screened Net Revenue",
    fontsize=15,
    fontweight="bold"
)
ax.set_xlabel("Month")
ax.set_ylabel("Net Revenue (₹)")
ax.yaxis.set_major_formatter(
    mtick.StrMethodFormatter("₹{x:,.0f}")
)

ax.tick_params(axis="x", rotation=60)
ax.grid(axis="y", linestyle="--", alpha=0.4)
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Step 24: Calculate month-over-month revenue changes

monthly_revenue["Previous_Month_Revenue_INR"] = (
    monthly_revenue["Net_Revenue_INR"].shift(1)
)

monthly_revenue["MoM_Change_INR"] = (
    monthly_revenue["Net_Revenue_INR"]
    - monthly_revenue["Previous_Month_Revenue_INR"]
)

monthly_revenue["MoM_Change_Percent"] = (
    monthly_revenue["MoM_Change_INR"]
    / monthly_revenue["Previous_Month_Revenue_INR"]
    * 100
)

print(
    monthly_revenue[
        [
            "Month",
            "Net_Revenue_INR",
            "MoM_Change_INR",
            "MoM_Change_Percent"
        ]
    ].round(2).to_string(index=False)
)
```

```python
# Step 25: Year-over-year revenue comparison
# Compare January–September only to avoid partial October 2026.

monthly_revenue["Year"] = (
    monthly_revenue["Month"].str[:4].astype(int)
)
monthly_revenue["Month_Number"] = (
    monthly_revenue["Month"].str[5:7].astype(int)
)

yoy_data = monthly_revenue[
    (monthly_revenue["Month_Number"] <= 9)
    & (monthly_revenue["Year"].isin([2025, 2026]))
]

yoy_comparison = (
    yoy_data
    .pivot(
        index="Month_Number",
        columns="Year",
        values="Net_Revenue_INR"
    )
    .reset_index()
)

yoy_comparison["Change_INR"] = (
    yoy_comparison[2026] - yoy_comparison[2025]
)

yoy_comparison["Change_Percent"] = (
    yoy_comparison["Change_INR"]
    / yoy_comparison[2025]
    * 100
)

print(yoy_comparison.round(2).to_string(index=False))
```

```python
# Step 26: Compare total January–September revenue

comparison_totals = (
    yoy_comparison[[2025, 2026]]
    .sum()
)

revenue_2025 = comparison_totals[2025]
revenue_2026 = comparison_totals[2026]

total_change = revenue_2026 - revenue_2025
total_change_percent = (total_change / revenue_2025) * 100

print("Jan–Sep 2025 revenue (INR):", round(revenue_2025, 2))
print("Jan–Sep 2026 revenue (INR):", round(revenue_2026, 2))
print("Absolute change (INR):", round(total_change, 2))
print("Overall YoY change (%):", round(total_change_percent, 2))
```

```python
# Step 27: Revenue contribution by product category

category_revenue = (
    eligible_sales
    .groupby("Category")
    .agg(
        Eligible_Sales_Rows=("Transaction_ID", "size"),
        Net_Revenue_INR=("Net_Line_Value_INR", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .reset_index()
    .sort_values("Net_Revenue_INR", ascending=False)
)

category_revenue["Revenue_Share_Percent"] = (
    category_revenue["Net_Revenue_INR"]
    / category_revenue["Net_Revenue_INR"].sum()
    * 100
)

print(category_revenue.round(2).to_string(index=False))
```

```python
# Step 30: Join eligible sales with product costs

eligible_with_cost = eligible_sales.merge(
    products_clean[
        ["Product_ID", "Unit_Cost_INR"]
    ],
    on="Product_ID",
    how="left",
    validate="many_to_one",
    indicator=True
)

print("Eligible sales rows:", len(eligible_sales))
print("Rows after product join:", len(eligible_with_cost))
print(
    "Rows with matched product cost:",
    (eligible_with_cost["_merge"] == "both").sum()
)
print(
    "Rows without matched product cost:",
    (eligible_with_cost["_merge"] == "left_only").sum()
)
print(
    "Missing unit costs:",
    eligible_with_cost["Unit_Cost_INR"].isna().sum()
)
```

```python
# Step 31: Calculate estimated gross profit

eligible_with_cost["Estimated_Cost_INR"] = (
    eligible_with_cost["Quantity"]
    * eligible_with_cost["Unit_Cost_INR"]
)

eligible_with_cost["Estimated_Gross_Profit_INR"] = (
    eligible_with_cost["Net_Line_Value_INR"]
    - eligible_with_cost["Estimated_Cost_INR"]
)

# Summarize profitability by product category
category_profit = (
    eligible_with_cost
    .groupby("Category")
    .agg(
        Net_Revenue_INR=("Net_Line_Value_INR", "sum"),
        Estimated_Cost_INR=("Estimated_Cost_INR", "sum"),
        Estimated_Gross_Profit_INR=(
            "Estimated_Gross_Profit_INR", "sum"
        )
    )
    .reset_index()
)

category_profit["Estimated_Gross_Margin_Percent"] = (
    category_profit["Estimated_Gross_Profit_INR"]
    / category_profit["Net_Revenue_INR"]
    * 100
)

category_profit = category_profit.sort_values(
    "Estimated_Gross_Profit_INR",
    ascending=False
)

print(category_profit.round(2).to_string(index=False))

print(
    "\nTotal estimated gross profit (INR):",
    round(eligible_with_cost["Estimated_Gross_Profit_INR"].sum(), 2)
)
```

```python
# Step 32: Validate total estimated gross profit and margin

total_net_revenue = eligible_with_cost[
    "Net_Line_Value_INR"
].sum()

total_estimated_cost = eligible_with_cost[
    "Estimated_Cost_INR"
].sum()

total_estimated_profit = eligible_with_cost[
    "Estimated_Gross_Profit_INR"
].sum()

overall_gross_margin = (
    total_estimated_profit / total_net_revenue
) * 100

print("Total net revenue (INR):", round(total_net_revenue, 2))
print("Total estimated cost (INR):", round(total_estimated_cost, 2))
print("Total estimated gross profit (INR):", round(total_estimated_profit, 2))
print("Overall estimated gross margin (%):", round(overall_gross_margin, 2))

print(
    "Profit reconciliation matches:",
    np.isclose(
        total_net_revenue - total_estimated_cost,
        total_estimated_profit
    )
)
```

```python
# Step 82: Summarize customer quality flags

customer_quality_columns = {
    "Age": "Age_Quality_Flag",
    "Gender": "Gender_Quality_Flag",
    "Mobile": "Mobile_Quality_Flag",
    "Email": "Email_Quality_Flag",
    "Signup Date": "Signup_Date_Quality_Flag",
    "Customer Segment": "Customer_Segment_Quality_Flag",
    "Loyalty Tier": "Loyalty_Tier_Quality_Flag",
    "City": "City_Quality_Flag",
    "State": "State_Quality_Flag",
    "Customer Name": "Customer_Name_Quality_Flag"
}

customer_quality_summary = []

for field_name, flag_column in customer_quality_columns.items():
    counts = customers_clean[flag_column].value_counts(dropna=False)

    for flag_value, count in counts.items():
        customer_quality_summary.append({
            "Field": field_name,
            "Quality_Status": flag_value,
            "Record_Count": int(count)
        })

customer_quality_summary = pd.DataFrame(customer_quality_summary)

print(customer_quality_summary.to_string(index=False))
```

```python
# Step 83: Summarize transaction quality flags

transaction_quality_columns = {
    "Transaction Date": "Date_Quality_Flag",
    "Customer ID": "Customer_ID_Quality_Flag",
    "Discount": "Discount_Quality_Flag",
    "Quantity": "Quantity_Quality_Flag",
    "Unit Price": "Price_Quality_Flag",
    "Transaction Status": "Transaction_Status_Quality_Flag",
    "Sales Channel": "Sales_Channel_Quality_Flag",
    "Payment Method": "Payment_Method_Quality_Flag"
}

transaction_quality_summary = []

for field_name, flag_column in transaction_quality_columns.items():
    if flag_column not in transactions_clean.columns:
        print(f"Missing flag column: {flag_column}")
        continue

    counts = transactions_clean[flag_column].value_counts(dropna=False)

    for flag_value, count in counts.items():
        transaction_quality_summary.append({
            "Field": field_name,
            "Quality_Status": flag_value,
            "Record_Count": int(count)
        })

transaction_quality_summary = pd.DataFrame(transaction_quality_summary)

print(transaction_quality_summary.to_string(index=False))
```

```python
# Step 84: Verify transaction quality flag columns

required_transaction_flags = [
    "Date_Quality_Flag",
    "Customer_ID_Quality_Flag",
    "Discount_Quality_Flag",
    "Quantity_Quality_Flag",
    "Price_Quality_Flag",
    "Transaction_Status_Quality_Flag",
    "Sales_Channel_Quality_Flag",
    "Payment_Method_Quality_Flag",
    "Has_Critical_Quality_Issue"
]

for column in required_transaction_flags:
    if column in transactions_clean.columns:
        print(f"{column}: Present")
    else:
        print(f"{column}: MISSING")
```
